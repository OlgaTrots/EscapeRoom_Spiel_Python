```python
import sys

class EscapeRoom:
    def __init__(self):
        self.zahl_1 = ""
        self.zahl_2 = ""
        self.tür_offen = False

    def p(self, *args, **kwargs):
        """Zuverlässiger Print mit erzwungenem Flush"""
        print(*args, **kwargs, flush=True)
        sys.stdout.flush()

    def info(self):
        self.p("================================================================")
        self.p(" ******** ESCAPE ROOM - DAS VERLASSENE HOTEL ******** ")
        self.p("================================================================")
        self.p()
        self.p("Du stehst in einer dunklen Halle eines verlassenen Hotels.")
        self.p("Vor dir ist eine schwere Holztür und überall stehen Bücherregale.")
        self.p("Es sieht aus wie eine alte Bibliothek. Es ist dunkel und die Tür ist verschlossen.")
        self.p("Um das Hotel zu verlassen, musst du drei Räume durchqueren.")
        self.p()
        self.p("Viel Spaß!")

    def abfrage(self, optionen):
        while True:
            self.p()
            for nummer, text in optionen.items():
                self.p(f"{nummer} - {text}")
            wahl = input("Deine Wahl: ").strip()
            if wahl in optionen:
                return wahl
            else:
                self.p("Ungültige Eingabe.")

    def beenden(self):
        self.p()
        self.p("Das spiel ist vorbei. Auf Wiedersehen!")
        self.p()
        sys.exit()

    def zimmer_1(self):
        self.p("================================================================")
        self.p("                           Zimmer 1")
        self.p("================================================================")
        self.p()
        self.p("Um weiterzukommen, musst du Hinweise finden.")
        self.p("Wo möchtest du anfangen?")
        self.p()
        optionen = {
            "1": "In einem Kerzenständer",
            "2": "In einer alten Vase auf dem Regal",
            "3": "Im Buch auf dem Tisch",
            "4": "Hinter dem Bild",
            "5": "Die Tür öffnen",
            "6": "Spiel beenden"
        }
        while True:
            wahl = self.abfrage(optionen)
            if wahl == "1":
                self.p()
                self.p("Hier gibt es nichts zu finden. Versuch es noch einmal!")
                self.p()
            elif wahl == "2":
                self.p()
                self.p("Hier gibt es nichts zu finden. Noch ein Versuch!")
                self.p()
            elif wahl == "3":
                self.p()
                self.p("\033[91m Richtig! Du hast etwas gefunden. Das ist einen alten Zettel.\033[0m")
                self.p("Auf dem Zettel steht eine Nummer 27. Merke dir die!")
                self.p()
                self.p("Schau dir an, was dort noch steht - es ist ein Rätsel.")
                self.p("Lies aufmerksam und versuch es zu lösen!")
                self.p()
                antwort = input("Was ist die Summe der ersten drei ungeraden Zahlen? ").strip()
                self.p()
                if antwort == "9":
                    self.p("\033[91m Richtig! Sieh mal, die Tür knarrt und öffnet sich ein Stück.\033[0m")
                    self.tür_offen = True
                    self.zahl_1 = "27"
                    self.p()
                    self.p("Jetzt kannst du die Tür öffnen und weitergehen.")
                    self.p()
                    break
                else:
                    self.p()
                    self.p("Leider falsch. Versuch es nochmal.")
                    self.p()
            elif wahl == "4":
                self.p()
                self.p("Hier ist nichts Verstecktes zu entdecken.")
                self.p()
            elif wahl == "5":
                if self.tür_offen:
                    self.p()
                    self.p("Du öffnest die Tür und gehst weiter.")
                    self.p()
                    break
                else:
                    self.p()
                    self.p("Die Tür ist noch verschlossen.")
                    self.p()
            elif wahl == "6":
                self.beenden()

    def zimmer_2(self):
        self.p("================================================================")
        self.p("                  Zimmer 2 - Rezeption")
        self.p("================================================================")
        self.p()
        self.p("Du bist an der Rezeption.")
        self.p("Auf dem Tisch liegen vier Zimmerschlüssel. Du musst einen auswählen.")
        optionen = {
            "1": "Zimmer 12",
            "2": "Zimmer 7",
            "3": "Zimmer 22",
            "4": "Zimmer 27",
            "6": "Spiel beenden"
        }
        self.p()
        while True:
            wahl = self.abfrage(optionen)
            if wahl in ["1", "2", "3"]:
                self.p()
                self.p("Dieser Schlüssel passt leider nicht.")
                self.p()
            elif wahl == "4":
                self.p()
                self.p("\033[91m Richtig!\033[0m")
                self.p()
                self.p("Am Schlüssel findest du eine Nummer.")
                self.p("Die Nummer ist 2026. Merke dir die!")
                self.zahl_2 = "2026"
                self.p()
                self.p("Jetzt kannst du die Tür öffnen und weitergehen.")
                self.p()
                break
            elif wahl == "6":
                self.beenden()

    def zimmer_3(self):
        self.p("================================================================")
        self.p("                     Zimmer 3 - Keller")
        self.p("================================================================")
        self.p()
        self.p("Du gehst die Treppe hinunter - Du bist im Keller.")
        self.p("Es ist feucht und dunkel hier. .... Und plötzlich hörst du ein Geräusch...")
        self.p("Was wirst du tun?")
        optionen = {
            "1": "Zurückgehen",
            "2": "Die Taschenlampe nehmen",
            "3": "Weitergehen",
            "4": "Still bleiben und sich nicht bewegen",
            "6": "Spiel beenden"
        }
        self.p()
        while True:
            wahl = self.abfrage(optionen)
            if wahl == "1":
                self.p()
                self.p("Nein, die Tür hinter dir ist bereits verschlossen.")
                self.p()
            elif wahl == "2":
                self.p()
                self.p("\033[91m Richtig! Du nimmst die Taschenlampe.\033[0m")
                self.p("Jetzt kannst du eine alte Tür mit Zahlenschloss sehen.")
                self.p()
                break
            elif wahl == "3":
                self.p()
                self.p("Das schaffst du nicht, es ist zu dunkel.")
                self.p()
            elif wahl == "4":
                self.p()
                self.p("Das wird dir nicht helfen, aus dem Keller zu entkommen.")
                self.p()
            elif wahl == "6":
                self.beenden()

    def ausgang(self):
        self.p()
        self.p("================================================================")
        self.p("                      Die letzte Tür")
        self.p("================================================================")
        self.p()
        self.p("Um das Zahlenschloss zu öffnen, brauchst du den Code.")
        self.p("Na los, versuch es!")

        summe = self.zahl_1 + self.zahl_2
        while True:
            code = input("Bitte schreib den Code: ").strip()
            if code == "6":
                self.beenden()
            elif code == summe:
                self.p()
                self.p("\033[91m         GLÜCKWÜNSCH! DU HAST ES GESCHAFFT!\033[0m")
                self.p("Die Tür öffnet sich langsam...")
                self.p("Endlich kannst du das Hotel verlassen.")
                self.p()
                break
            else:
                self.p("Falscher Code. Versuche es noch einmal.")

    def starten(self):
        self.info()
        self.zimmer_1()
        self.zimmer_2()
        self.zimmer_3()
        self.ausgang()



spiel = EscapeRoom()
spiel.starten()
     
```
